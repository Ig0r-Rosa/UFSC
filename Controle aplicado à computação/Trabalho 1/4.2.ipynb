```python
import control
import matplotlib.pyplot as plt
```

```python
## 12 / (s² + 2s + 10)
G = control.tf([4], [1, 2, 20])
```

```python
# Resposta ao degrau em malha aberta
t_ol, y_ol = control.step_response(G)
plt.figure()
plt.plot(t_ol, y_ol)
plt.title("Resposta ao degrau - malha aberta")
plt.xlabel("Tempo (s)")
plt.ylabel("Saída")
plt.grid(True)
```

```python
# Cálculo do erro em regime estacionário
# Valor final da saída (usando Teorema do Valor Final)
y_final = y_ol[-1]  # Último valor da resposta ao degrau

# Ganho DC (valor em DC = s→0)
G_dc = control.dcgain(G)
print(f"Ganho DC da função de transferência: {G_dc}")

# Valor esperado em regime estacionário para entrada degrau unitário
print(f"\nValor da saída em regime estacionário: {y_final:.4f}")

# Erro em regime estacionário
erro_ss = 1 - y_final
print(f"Erro em regime estacionário (1 - saída): {erro_ss:.4f}")
print(f"\nComo esperado, o erro é aproximadamente 0.8")

# Adicionar uma linha no gráfico mostrando o valor final
plt.figure()
plt.plot(t_ol, y_ol, linewidth=2, label='Saída')
plt.axhline(y=y_final, color='r', linestyle='--', label=f'Valor final = {y_final:.2f}')
plt.axhline(y=1, color='g', linestyle='--', label='Valor desejado = 1.0')
plt.title("Resposta ao degrau unitário - Malha aberta")
plt.xlabel("Tempo (s)")
plt.ylabel("Saída")
plt.legend()
plt.grid(True)
plt.ylim([0, 1.2])
plt.show()

print(f"\n{'='*50}")
print("RESULTADO DO EXERCÍCIO PC4.2")
print(f"{'='*50}")
print(f"Função de transferência: G(s) = 4/(s² + 2s + 20)")
print(f"Entrada: Degrau unitário")
print(f"Valor desejado em regime estacionário: 1")
print(f"Valor obtido em regime estacionário: {y_final:.1f}")
print(f"Erro em regime estacionário: {erro_ss:.1f}")
print(f"{'='*50}")
```
