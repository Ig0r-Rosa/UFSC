```python
import control
import matplotlib.pyplot as plt
```

```python
## 12 / (s² + 2s + 10)
G = control.tf([12], [1, 2, 10])
```

```python
# Resposta ao degrau em malha aberta
t_ol, y_ol = control.step_response(G)
plt.figure()
plt.plot(t_ol, y_ol)
plt.title("Resposta ao degrau - malha aberta")
plt.xlabel("Tempo (s)")
plt.ylabel("Saída")
plt.grid(True)
```

```python
# Obtendo as informações da resposta ao degrau
info = control.step_info(G)
```

```python
# Máxima ultrapassagem percentual
mp = info['Overshoot']
print(f"Máxima Ultrapassagem (MP): {mp:.2f}%")
```

```python
# Valor final de regime (Steady State Value)
y_final = control.dcgain(G)
print(f"Valor final (y_ss): {y_final:.2f}")
```

```python
# Erro em regime estacionário (para entrada degrau unitário)
ess = 1 - y_final
print(f"Erro em regime estacionário (e_ss): {ess:.2f}")
```
